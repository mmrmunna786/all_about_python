```python
df1 = pandas.read_csv( "traffic.csv" )
df1
```

```python
specific_data = df1.loc[ 2:5, "Country":"Clicks" ]
specific_data
```

selecting rows from 6-9 and columns visitors-clicks using "iloc" attribute/method

update/add columns and rows

```python
# adding a new column to the dataframe
df1["New Column"] = [ "A" ] * 10
df1
```
